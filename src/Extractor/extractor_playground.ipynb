Process tracks from **custom** and **M-DJCUE** datasets that have valid cue point annotations (at least 1 entry + 1 exit cue)

```python
# [CLEANUP] Remove Empty/Invalid M-DJCUE Annotations
# This script scans the M-DJCUE annotations and removes files that don't have valid cue points.
# Use this to clean up bad copies created during matching attempts.

import os
import json
from pathlib import Path

def cleanup_invalid_annotations():
    print("Cleaning up invalid M-DJCUE annotations...")
    print("="*60)
    
    ann_dir = Path("../../data/M-DJCUE/annotations")
    
    if not ann_dir.exists():
        print(f"Directory not found: {ann_dir}")
        return

    jams_files = list(ann_dir.glob("*.jams"))
    print(f"Scanning {len(jams_files)} JAMS files...")
    
    removed_count = 0
    kept_count = 0
    
    for jams_file in jams_files:
        is_valid = False
        try:
            with open(jams_file, 'r') as f:
                data = json.load(f)
            
            # Check for valid cues
            has_entry = False
            has_exit = False
            has_nonzero = False
            
            for annotation in data.get('annotations', []):
                if annotation.get('namespace') == 'cue_point':
                    for point in annotation.get('data', []):
                        time_val = point.get('time', -1)
                        label = point.get('value', {}).get('label', '')
                        
                        if time_val >= 0:
                            if label == 'IN':
                                has_entry = True
                            elif label == 'OUT':
                                has_exit = True
                            
                            if time_val > 0:
                                has_nonzero = True
            
            # Validity criteria: Must have IN+OUT and at least one non-zero timestamp
            if has_entry and has_exit and has_nonzero:
                is_valid = True
                
        except Exception as e:
            print(f"Error reading {jams_file.name}: {e}")
            is_valid = False
            
        if not is_valid:
            try:
                os.remove(jams_file)
                removed_count += 1
                # print(f"Removed invalid: {jams_file.name}")
            except Exception as e:
                print(f"Failed to remove {jams_file.name}: {e}")
        else:
            kept_count += 1
            
    print("-" * 60)
    print(f"Cleanup Complete:")
    print(f"   Removed: {removed_count} invalid files")
    print(f"   Kept:    {kept_count} valid files")
    print("="*60)

cleanup_invalid_annotations()
```

```python
import json
import time
import re
import difflib
from pathlib import Path
from tqdm.auto import tqdm

# Define both datasets to process
DATASETS = [
    {
        'name': 'custom',
        'audio_dir': Path("../../data/custom/audio"),
        'annotation_dir': Path("../../data/custom/annotations"),
        'output_dir': Path("../../data/processed")
    },
    {
        'name': 'M-DJCUE',
        'audio_dir': Path("../../data/M-DJCUE/audio"),
        'annotation_dir': Path("../../data/M-DJCUE/annotations"),
        'output_dir': Path("../../data/processed")
    }
]

print("Scanning for tracks with valid annotations from BOTH datasets...")
print("=" * 70)

# Find tracks with valid cue points
def normalize_filename(filename):
    """Aggressively normalize filename for fuzzy matching"""
    # Remove extension
    name = filename.replace('.mp3', '').replace('.jams', '').lower()
    
    # Remove year prefixes like (2012) or v_ prefix
    name = re.sub(r'^\(\d{4}\)', '', name)
    name = re.sub(r'^v_', '', name)
    
    # Remove content between brackets/parens (metadata/mix info often differs)
    # e.g. "Song (Original Mix)" vs "Song [Original]"
    name = re.sub(r'\[.*?\]', '', name)
    name = re.sub(r'\(.*?\)', '', name)
    
    # Strip non-alphanumeric chars (keep only letters/numbers)
    name = re.sub(r'[^a-z0-9]', '', name)
    
    return name

def find_matching_annotation(audio_file, annotation_dir):
    """Find best matching annotation with detailed score"""
    audio_stem = audio_file.stem
    all_jams = list(annotation_dir.glob("*.jams"))
    
    # 1. Exact Match
    exact = annotation_dir / f"{audio_stem}.jams"
    if exact.exists(): 
        return exact, 1.0, "Exact"
    
    # 2. 'v_' Prefix Match
    v_match = annotation_dir / f"v_{audio_stem}.jams"
    if v_match.exists(): 
        return v_match, 0.99, "v_prefix"
    
    # 3. Fuzzy Similarity Match
    audio_norm = normalize_filename(audio_stem)
    # Create simple token set for robust matching (e.g. reordered words)
    audio_tokens = set(audio_norm.split())

    best_match = None
    best_score = 0.0
    match_method = "None"
    
    for jam in all_jams:
        jam_norm = normalize_filename(jam.stem)
        
        # Score A: Sequence Ratio (Order sensitive, good for Typos)
        seq_score = difflib.SequenceMatcher(None, audio_norm, jam_norm).ratio()
        
        # Score B: Token Set IO (Order independent, good for Reordering)
        jam_tokens = set(jam_norm.split())
        if audio_tokens and jam_tokens:
            intersection = audio_tokens.intersection(jam_tokens)
            union = audio_tokens.union(jam_tokens)
            token_score = len(intersection) / len(union)
        else:
            token_score = 0.0
            
        # Use the better of the two scores
        score = max(seq_score, token_score)
        
        if score > best_score:
            best_score = score
            best_match = jam
            match_method = "Seq" if seq_score >= token_score else "Token"
            
    # Threshold: Match must be > 75% similar to be trusted
    if best_score > 0.75:
        return best_match, best_score, f"Fuzzy({match_method})"
        
    return None, best_score, "Failed"

def check_valid_cues(jams_file):
    """Check if JAMS file has proper annotations (not all zeros)"""
    try:
        with open(jams_file, 'r') as f:
            data = json.load(f)
        
        entry_cues = []
        exit_cues = []
        
        for annotation in data.get('annotations', []):
            if annotation.get('namespace') == 'cue_point':
                for point in annotation.get('data', []):
                    time_val = point.get('time', -1)
                    label = point.get('value', {}).get('label', '')
                    
                    if time_val >= 0:  # Accept all valid times including 0.0
                        if label == 'IN':
                            entry_cues.append(time_val)
                        elif label == 'OUT':
                            exit_cues.append(time_val)
        
        has_entry = len(entry_cues) > 0
        has_exit = len(exit_cues) > 0
        has_nonzero = any(t > 0 for t in entry_cues + exit_cues)
        
        is_valid = has_entry and has_exit and has_nonzero
        return is_valid, len(entry_cues), len(exit_cues)
    
    except Exception as e:
        return False, 0, 0

# Scan all datasets
valid_tracks = []

for dataset in DATASETS:
    dataset_name = dataset['name']
    audio_dir = dataset['audio_dir']
    annotation_dir = dataset['annotation_dir']
    output_dir = dataset['output_dir']
    
    # Create output directory if it doesn't exist
    output_dir.mkdir(parents=True, exist_ok=True)
    
    print(f"\n{dataset_name}:")
    print(f"{'Audio File':<40} | {'Score':<5} | {'Status':<10} | {'Match Info'}")
    print("-" * 100)
    
    # Get all audio files
    all_audio_files = sorted(list(audio_dir.glob("*.mp3")))
    
    # Track statistics
    dataset_with_annotation = 0
    dataset_valid_count = 0
    dataset_rejected = 0
    
    for audio_file in all_audio_files:
        # Use smart matching function with score
        annotation_file, score, method = find_matching_annotation(audio_file, annotation_dir)
        
        status = "❌"
        info = f"Best: {score:.2f}"
        
        if annotation_file is not None:
            dataset_with_annotation += 1
            is_valid, num_entry, num_exit = check_valid_cues(annotation_file)
            
            if is_valid:
                status = "✅"
                info = f"{method} ({score:.2f})"
                valid_tracks.append({
                    'audio_file': audio_file,
                    'annotation_file': annotation_file,
                    'output_dir': output_dir,
                    'dataset': dataset_name,
                    'num_entry': num_entry,
                    'num_exit': num_exit
                })
                dataset_valid_count += 1
            else:
                status = "⚠️" # Matched but invalid cues
                info = f"Invalid Cues ({num_entry}/{num_exit})"
                dataset_rejected += 1
        
        # Print short summary for every file
        short_name = (audio_file.name[:37] + '..') if len(audio_file.name) > 37 else audio_file.name
        print(f"{short_name:<40} | {score:.2f}  | {status:<10} | {info}")

    print("-" * 100)
    print(f"Summary for {dataset_name}:")
    print(f"   Audio files: {len(all_audio_files)}")
    print(f"   Matches Found: {dataset_with_annotation}")
    print(f"   Valid Tracks: {dataset_valid_count}")
    print(f"   Invalid/Rejected: {dataset_rejected}")

print(f"\n{'='*70}")
print(f"TOTAL: {len(valid_tracks)} tracks with valid annotations across all datasets")
print(f"{'='*70}")
```

```python

import difflib

def list_unmatched_files():
    print("Finding unmatched files in M-DJCUE...")
    print("="*70)
    
    audio_dir = Path("../../data/M-DJCUE/audio")
    ann_dir = Path("../../data/M-DJCUE/annotations")
    
    all_mp3s = sorted(list(audio_dir.glob("*.mp3")))
    all_jams = sorted(list(ann_dir.glob("*.jams")))
    
    # Track used annotations to find orphans
    used_jams = set()
    unmatched_mp3s = []
    
    # 1. Simulate matching process
    for mp3 in all_mp3s:
        jam, score, method = find_matching_annotation(mp3, ann_dir)
        if jam and score > 0.75:
            used_jams.add(jam)
        else:
            unmatched_mp3s.append(mp3)
            
    # 2. Find unused annotations
    unused_jams = [j for j in all_jams if j not in used_jams]
    
    # 3. Print Report
    print(f"\n--- UNMATCHED MP3 FILES ({len(unmatched_mp3s)}) ---")
    for mp3 in unmatched_mp3s:
        print(f"{mp3.name}")

    print(f"\n--- UNUSED ANNOTATION FILES ({len(unused_jams)}) ---")
    for jam in unused_jams:
        print(f"{jam.name}")
        
    print(f"\n{'='*70}")
    print("SUGGESTION: Rename one of the unused JAMS files to match an unmatched MP3.")
    print("Example: Copy 'Song [Original].jams' -> 'Song (Original Mix).jams' (if that is the MP3 name)")

list_unmatched_files()
```

## Process and pickle custom files

```python
import pickle
import sys
from pathlib import Path
from tqdm.auto import tqdm

# Ensure src is in path for imports
# Adding the project root to path so we can import modules
sys.path.append(str(Path.cwd().parent))     # Adds src

from extractor import Extractor # Import directly from the file in the same directory

# Setup Paths
AUDIO_DIR = Path("../../data/custom/audio")
ANNOTATION_DIR = Path("../../data/custom/annotations")
OUTPUT_DIR = Path("../../data/processed")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Helpers from previous cells (re-defined for standalone use if needed) 
def normalize_filename(filename):
    """Normalize filename for matching"""
    name = filename.replace('.mp3', '').replace('.jams', '').lower()
    name = re.sub(r'^\(\d{4}\)', '', name)
    name = re.sub(r'^v_', '', name)
    name = re.sub(r'\[.*?\]', '', name)
    name = re.sub(r'\(.*?\)', '', name)
    name = re.sub(r'[^a-z0-9]', '', name)
    return name

def find_annotation(audio_path, annotation_dir):
    """Find corresponding JAMS file"""
    audio_stem = audio_path.stem
    
    # Exact match
    exact = annotation_dir / f"{audio_stem}.jams"
    if exact.exists():
        return exact

    # Fuzzy match fallback
    audio_norm = normalize_filename(audio_stem)
    best_match = None
    best_score = 0.0
    
    for jam in annotation_dir.glob("*.jams"):
        jam_norm = normalize_filename(jam.stem)
        score = difflib.SequenceMatcher(None, audio_norm, jam_norm).ratio()
        if score > best_score:
            best_score = score
            best_match = jam
            
    if best_score > 0.8: # Threshold
        return best_match
    return None

# Main Processing Loop
print(f"Starting extraction for Custom Manual Annotations...")
print(f"Audio Source: {AUDIO_DIR}")
print(f"Annotations: {ANNOTATION_DIR}")
print(f"Output: {OUTPUT_DIR}")

# Pass vocal separation setting if needed (default false for speed)
extractor = Extractor(enable_vocal_separation=False)
audio_files = list(AUDIO_DIR.glob("*.mp3"))
processed_count = 0
skipped_count = 0

for audio_path in tqdm(audio_files, desc="Processing Tracks"):
    try:
        # Find matching annotation
        annotation_path = find_annotation(audio_path, ANNOTATION_DIR)
        
        if not annotation_path:
            print(f"⚠️ No annotation found for: {audio_path.name}")
            skipped_count += 1
            continue
            
        print(f"Processing: {audio_path.name} -> {annotation_path.name}")
        
        # Extract Features
        # Using track_id as stem
        track = extractor.extract(
            audio_path=audio_path,
            annotation_path=annotation_path,
            track_id=audio_path.stem,
            source="custom"
        )
        
        # Save validation
        if not track.has_cue_labels:
            print(f"⚠️ Track {track.track_id} has no cue labels extracted!")
            
        # Serialize to Pickle
        output_pickle = OUTPUT_DIR / f"{track.track_id}.pkl"
        with open(output_pickle, 'wb') as f:
            pickle.dump(track, f)
            
        processed_count += 1
        
    except Exception as e:
        print(f"Error processing {audio_path.name}: {e}")
        skipped_count += 1

print("\n" + "="*50)
print(f"Processing Complete!")
print(f"Successfully processed: {processed_count}")
print(f"Skipped/Failed: {skipped_count}")
print(f"Output saved to: {OUTPUT_DIR}")
print("="*50)
```
